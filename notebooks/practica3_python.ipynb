El inventario de tres almacenes se representa como una lista de listas. Cada sublista sigue el formato `[codigo_almacen, producto, unidades, coste_unitario]`.

```python
inventario = [
    ["ALM-NORTE", "Robot Aspirador", 34, 201.50],
    ["ALM-NORTE", "Monitor 27", 58, 142.00],
    ["ALM-SUR",   "Silla Ergonomica", 12, 178.90],
    ["ALM-SUR",   "Robot Aspirador", 7, 201.50],
    ["ALM-ESTE",  "Cafetera Expres", 41, 131.20],
    ["ALM-ESTE",  "Monitor 27", 25, 142.00],
]
```

**Se pide:**

1. Imprime el nombre del producto de la **cuarta** fila usando doble indexación.
2. Crea una nueva lista `valor_stock` donde cada elemento sea `[producto, unidades * coste_unitario]` redondeado a dos decimales. Resuélvelo con un bucle `for` sobre `inventario`.
3. Calcula el **valor inmovilizado total** del inventario.
4. Extrae, mediante *slicing*, una sublista con solo las filas del `ALM-SUR` (usa el hecho de que están contiguas) y calcula su valor.
5. Explica en una celda Markdown de **dos frases** qué problema aparece si dos almacenes dejan de estar contiguos y por qué el *slicing* no es una estrategia robusta de filtrado.

**Comprobación:** el valor inmovilizado total debe ser 27 573,50 € y el del `ALM-SUR`, 3 557,30 €.

````python
inventario = [
    ["ALM-NORTE", "Robot Aspirador", 34, 201.50],
    ["ALM-NORTE", "Monitor 27", 58, 142.00],
    ["ALM-SUR",   "Silla Ergonomica", 12, 178.90],
    ["ALM-SUR",   "Robot Aspirador", 7, 201.50],
    ["ALM-ESTE",  "Cafetera Expres", 41, 131.20],
    ["ALM-ESTE",  "Monitor 27", 25, 142.00],
]

productoCuartaFila = inventario[3][1]
print(f"Producto de la 4ª fila: {productoCuartaFila}")

valorStock = []
for fila in inventario:
    producto = fila[1]
    unidades = fila[2]
    costeUnitario = fila[3]
    valor = round(unidades * costeUnitario, 2)
    valorStock.append([producto, valor])

valorTotal = sum(fila[1] for fila in valorStock)
print(f"Valor inmovilizado total: {valorTotal} €")

inventarioSur = inventario[2:4]
valorSur = round(sum(fila[2] * fila[3] for fila in inventarioSur), 2)
print(f"Valor del ALM-SUR: {valorSur} €")

print("\n Comprobacion")
print(f"¿El valor total es 27573.5?: {valorTotal == 27573.5}")
print(f"¿El valor SUR es 3557.3?: {valorSur == 3557.3}")
````

**Explicacion:**
El uso de slicing asume obligatoriamente que los registros están ordenados y agrupados en bloques contiguos predecibles. Si el inventario se desordena o se añaden nuevos registros intercalados, el *slicing* basado en posiciones capturará datos mezclados y erróneos, por lo que usar condiciones lógicas de filtrado (ej: `if fila[0] == "ALM-SUR"`) es la única estrategia realmente robusta.

### Ejercicio 5 — Funciones propias para el cálculo de importes

**Se pide:**

1. Define la función `calcular_importe(unidades, precio_unitario, descuento_pct=0)` que devuelva el importe neto redondeado a dos decimales. El parámetro `descuento_pct` debe tener valor por defecto.
2. Define `resumen_pedido(unidades, precio_unitario, descuento_pct=0)` que devuelva **una tupla de tres elementos**: `(importe_bruto, ahorro, importe_neto)`.
3. Llama a `resumen_pedido` usando **argumentos por palabra clave** y desempaqueta el resultado en tres variables.
4. Documenta ambas funciones con *docstring* y muestra la ayuda con `help(calcular_importe)`.
5. Usa `max()`, `min()`, `sorted()` y `len()` sobre la lista `importes` que obtengas al aplicar `calcular_importe` a estos cinco pedidos:
    
    ```python
    pedidos = [
        (3, 59.90, 10),
        (1, 279.00, 0),
        (5, 18.50, 20),
        (2, 149.00, 5),
        (4, 89.90, 15),
    ]
    ```
    
    **Comprobación:** `calcular_importe(3, 59.90, 10)` debe devolver `161.73`.

````python
def calcular_importe(unidades, precio_unitario, descuento_pct=0):
    """
    Calcula el importe neto de un pedido aplicando el descuento indicado.
    Devuelve el valor final redondeado a dos decimales.
    """
    importeBruto = unidades * precio_unitario
    descuento = importeBruto * (descuento_pct / 100)
    return round(importeBruto - descuento, 2)

def resumen_pedido(unidades, precio_unitario, descuento_pct=0):
    """
    Calcula los totales de una línea de pedido.
    Devuelve una tupla con el importe bruto, el ahorro y el importe neto.
    """
    importeBruto = unidades * precio_unitario
    ahorro = importeBruto * (descuento_pct / 100)
    importeNeto = importeBruto - ahorro
    return (round(importeBruto, 2), round(ahorro, 2), round(importeNeto, 2))

importeBruto, ahorro, importeNeto = resumen_pedido(unidades=3, precio_unitario=59.90, descuento_pct=10)
print(f"Bruto: {importeBruto} | Ahorro: {ahorro} | Neto: {importeNeto}\n")

help(calcular_importe)

pedidos = [
    (3, 59.90, 10),
    (1, 279.00, 0),
    (5, 18.50, 20),
    (2, 149.00, 5),
    (4, 89.90, 15),
]

listaImportes = []
for p in pedidos:
    listaImportes.append(calcular_importe(p[0], p[1], p[2]))

print(f"\nImporte máximo: {max(listaImportes)}")
print(f"Importe mínimo: {min(listaImportes)}")
print(f"Importes ordenados: {sorted(listaImportes)}")
print(f"Total de importes calculados: {len(listaImportes)}")

print(f"\nComprobación: {calcular_importe(3, 59.90, 10)}")
````

### Ejercicio 11 — Estadística descriptiva sobre telemetría simulada

**Se pide:**

1. Genera con `rng = np.random.default_rng(2024)` dos arrays de 500 elementos: `temperatura` con `rng.normal(68, 6.5, 500)` y `vibracion` con `np.abs(rng.normal(2.4, 0.8, 500))`.
2. Calcula media, mediana, desviación típica, mínimo y máximo de `temperatura`. Compara media y mediana y razona en una frase qué indica la diferencia.
3. Calcula los percentiles 25, 50, 75 y 95 de `vibracion` con `np.percentile()`.
4. Apila ambos arrays en una matriz de forma `(500, 2)` con `np.column_stack()` y verifica el `shape`.
5. Calcula la matriz de correlación con `np.corrcoef()` y explica en una frase si existe relación lineal entre temperatura y vibración en estos datos simulados.
6. Define `umbral = media + 2 * desviación` de la temperatura, cuenta cuántas lecturas lo superan y calcula qué porcentaje del total representan.
7. Explica en dos frases por qué fijar la semilla (`default_rng(2024)`) es imprescindible para que un análisis sea reproducible por un tercero.

````python
import numpy as np

rng = np.random.default_rng(2024)
temperatura = rng.normal(68, 6.5, 500)
vibracion = np.abs(rng.normal(2.4, 0.8, 500))

mediaTemp = np.mean(temperatura)
medianaTemp = np.median(temperatura)
desviacionTemp = np.std(temperatura)
minimoTemp = np.min(temperatura)
maximoTemp = np.max(temperatura)

print(mediaTemp)
print(medianaTemp)
print(desviacionTemp)
print(minimoTemp)
print(maximoTemp)

percentilesVibracion = np.percentile(vibracion, [25, 50, 75, 95])
print(percentilesVibracion)

matrizApilada = np.column_stack((temperatura, vibracion))
print(matrizApilada.shape)

matrizCorrelacion = np.corrcoef(temperatura, vibracion)
print(matrizCorrelacion)

umbral = mediaTemp + 2 * desviacionTemp
lecturasSuperadas = np.sum(temperatura > umbral)
porcentajeSuperadas = (lecturasSuperadas / len(temperatura)) * 100

print(lecturasSuperadas)
print(porcentajeSuperadas)
````

**Explicacion sobre media y mediana:** La similitud casi exacta entre la media y la mediana indica que la distribución de las temperaturas es simétrica (típicamente una campana de Gauss normal), sin que existan valores atípicos extremos que sesguen la media hacia un lado.

**Explicacion sobre la correlación:** Al observar un valor muy cercano a 0 en las posiciones fuera de la diagonal principal de la matriz (aprox. -0.06), se concluye que no existe ninguna relación lineal entre la temperatura y la vibración en estos datos simulados.

**Explicacion sobre la semilla:** Fijar la semilla obliga al algoritmo a iniciar siempre desde el mismo estado matemático exacto para generar los números pseudoaleatorios. Esto es imprescindible porque garantiza que un tercero obtendrá exactamente el mismo conjunto de datos y los mismos resultados al reproducir el código, asegurando la validez del análisis.
